## Ingresar datos iniciales del modelo

para este modelo se usaran los siguientes datos iniciales:

- C = 100
- lambda1 = 30
- lambda2 = 20
- theta1 = 1
- theta2 = 2


```python
C = 100
lambda1 = 30
lambda2 = 20
theta1 = 1
theta2 = 2

```

luego procedemos a calcular las cargas $\rho_1$ $\rho_2$ y leugo las cargas normalizadas $a_1$ $a_2$  y por ultimo calcular la carga normalizada total $a$

```python
rho1 = lambda1 * theta1
rho2 = lambda2 * theta2

a1 = rho1 / C
a2 = rho2 / C

a = a1 + a2
```

comprobamos la estabilidad de la estabilidad del sistema

```python
if a < 1:
    print("The system is stable")
else:
    raise Exception("The system is unstable")
```

 los estados del sistema  serar guardadas en tuplas

ahora se calculara la probabilidad del sistema vacio o $P(0,0)$.
como no se pueden calcular un numero infinito de estados del sistema, se usara una aproximacion la cual estara limitada a $n_1 + n_2 <= 40$ donde $P(n_1, n_2)$

```python
from math import factorial

K = 40
suma = 0

for n1 in range(K + 1):
    for n2 in range(K - n1 + 1):
        peso = (
            factorial(n1 + n2)
            * a1**n1
            * a2**n2
            / (factorial(n1) * factorial(n2))
        )
        suma += peso

p00 = 1 / suma

print("p(0,0) ≈", p00)


```

ahora crearemos un diccionario en la que guardaremos las probababilidades obtenidas y luego las calcularemos

```python
probabilities = {}
probabilities[(0, 0)] = p00

for n1 in range(K + 1):
    for n2 in range(K-n1 + 1):
        probabilities[(n1, n2)] =( (a1 ** n1 * a2 **n2 )/  ( factorial(n1)  * factorial(n2) )   ) * factorial(n1+n2) * p00


display(probabilities)
```

ahora calculamos los estados de bloqueo, como en este modelo $E_1$ = $E_2$ = 0, la probabilidad en este caso es 0 

```python
E1 = 0
E2= 0
```

tambien calculamos los valores medios de solcitudes respondidas

```python
N1 = lambda1 * theta1 / ((theta1* lambda1) + (theta2 * lambda2))
N2 = lambda2 * theta2 / ((theta1* lambda1) + (theta2 * lambda2))

print("N1 =", N1)
print("N2 =", N2)
```

por ultimo calculamos el tiempo medio del servicio

```python
T1 = N1 / lambda1
T2 = N2 / lambda2

print("T1 =", T1)
print("T2 =", T2)
```

en conclusion estos serian los datos

```python
print("states probabilities:")

for i in probabilities:
    if i[0] + i[1] == K:
        print("  p(", i[0], ",", i[1], ") =", probabilities[i])

print("\nblocking probabilities:")

print(" E1 =", E1)
print(" E2 =", E2)

print("\nСреднее число N_i")
print(" N1 =", N1)
print(" N2 =", N2)

print("\nСреднее время T_i")
print(" T1 =", T1)
print(" T2 =", T2)
```
